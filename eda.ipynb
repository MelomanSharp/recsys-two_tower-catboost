## RFM-analysis

```python
latest_date = transactions['t_dat'].max()

rfm = transactions.groupby('customer_id').agg({
    't_dat': lambda x: (latest_date - x.max()).days, # Recency
    'article_id': 'count',                           # Frequency
    'price': 'sum'                                   # Monetary
}).rename(columns={'t_dat': 'recency', 'article_id': 'frequency', 'price': 'monetary'})

# Single Segmentation by Recency
rfm['segment'] = pd.qcut(rfm['recency'], 3, labels=['Active', 'Slipping', 'Dormant'])

plt.figure(figsize=(8, 4))
sns.countplot(data=rfm, x='segment', palette='viridis')
plt.title('User Segmentation by Recency (RFM)')
plt.show()
```

import numpy as np

def calculate_psi(expected, actual, bins=20):
    expected = expected[~np.isnan(expected) & np.isfinite(expected)]
    actual = actual[~np.isnan(actual) & np.isfinite(actual)]
    
    if len(expected) == 0 or len(actual) == 0:
        return np.nan
    
    bins_edges = np.histogram_bin_edges(expected, bins=bins)
    expected_pct = np.histogram(expected, bins=bins_edges)[0] / len(expected)
    actual_pct = np.histogram(actual, bins=bins_edges)[0] / len(actual)
    
    # Защита от деления на ноль и log(0)
    expected_pct = np.where(expected_pct == 0, 1e-6, expected_pct)
    actual_pct = np.where(actual_pct == 0, 1e-6, actual_pct)
    
    psi = np.sum((actual_pct - expected_pct) * np.log(actual_pct / expected_pct))
    return psi if np.isfinite(psi) else np.nan

transactions['t_dat'] = pd.to_datetime(transactions['t_dat'])
transactions['year_month'] = transactions['t_dat'].dt.to_period('M')
months = sorted(transactions['year_month'].unique())

print(f"Data range: {months[0]} to {months[-1]}")
print(f"Total months: {len(months)}")

psi_price = []
psi_popularity = []
psi_channel = []

for i in range(1, len(months)):
    prev_month = transactions[transactions['year_month'] == months[i-1]]
    curr_month = transactions[transactions['year_month'] == months[i]]
    
    if len(prev_month) < 100 or len(curr_month) < 100:
        continue
    
    psi_p = calculate_psi(prev_month['price'].values, curr_month['price'].values, bins=20)
    psi_price.append({'month': str(months[i]), 'psi': psi_p})
    
    prev_pop = prev_month.groupby('article_id').size()
    curr_pop = curr_month.groupby('article_id').size()
    
    common_items = prev_pop.index.intersection(curr_pop.index)
    if len(common_items) > 10:
        psi_pop = calculate_psi(prev_pop.loc[common_items].values.astype(float), 
                                curr_pop.loc[common_items].values.astype(float), 
                                bins=20)
        psi_popularity.append({'month': str(months[i]), 'psi': psi_pop})
    
    prev_ch_dist = prev_month['sales_channel_id'].value_counts(normalize=True)
    curr_ch_dist = curr_month['sales_channel_id'].value_counts(normalize=True)
    
    all_channels = prev_ch_dist.index.union(curr_ch_dist.index)
    prev_ch = prev_ch_dist.reindex(all_channels, fill_value=1e-6).values
    curr_ch = curr_ch_dist.reindex(all_channels, fill_value=1e-6).values
    
    psi_ch = np.sum((curr_ch - prev_ch) * np.log(curr_ch / prev_ch))
    if np.isfinite(psi_ch):
        psi_channel.append({'month': str(months[i]), 'psi': psi_ch})

psi_price_df = pd.DataFrame(psi_price)
psi_popularity_df = pd.DataFrame(psi_popularity)
psi_channel_df = pd.DataFrame(psi_channel)

fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# Price PSI
if not psi_price_df.empty:
    sns.lineplot(data=psi_price_df, x='month', y='psi', marker='o', ax=axes[0], color='blue')
axes[0].axhline(y=0.25, color='r', linestyle='--', label='Retrain Threshold (0.25)')
axes[0].axhline(y=0.1, color='g', linestyle='--', label='Minor Drift (0.1)')
axes[0].set_title('PSI: Price Distribution')
axes[0].set_ylabel('PSI Value')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=45)

# Popularity PSI
if not psi_popularity_df.empty:
    sns.lineplot(data=psi_popularity_df, x='month', y='psi', marker='o', ax=axes[1], color='orange')
axes[1].axhline(y=0.25, color='r', linestyle='--', label='Retrain Threshold (0.25)')
axes[1].axhline(y=0.1, color='g', linestyle='--', label='Minor Drift (0.1)')
axes[1].set_title('PSI: Item Popularity Distribution')
axes[1].set_ylabel('PSI Value')
axes[1].legend()
axes[1].tick_params(axis='x', rotation=45)

# Channel PSI
if not psi_channel_df.empty:
    sns.lineplot(data=psi_channel_df, x='month', y='psi', marker='o', ax=axes[2], color='purple')
axes[2].axhline(y=0.25, color='r', linestyle='--', label='Retrain Threshold (0.25)')
axes[2].axhline(y=0.1, color='g', linestyle='--', label='Minor Drift (0.1)')
axes[2].set_title('PSI: Sales Channel Distribution')
axes[2].set_ylabel('PSI Value')
axes[2].legend()
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()

### Drift Analysis: Honest Conclusions

The 0.25 threshold is an industrial heuristic, but for our dataset:

1. The price distribution is stable (PSI < 0.25 over the entire horizon).

In general, it's expected result: the price at H&M is normalized [0, 1] and not subject to seasonal fluctuations.
→ Retraining by price PSI will not be performed. This is not a bug, because price is simply not a driver of drift in this dataset.

2. The largest drift is observed in September 2019 (PSI ≈ 0.17) –
the beginning of the fall season. This is a moderate drift (0.1–0.25), which warrants monitoring,
but not automatic retraining.

3. So obviously, for H&M's production system, the 0.25 price threshold is too conservative.

If we want to capture seasonal shifts, we need to:

- either lower the threshold to 0.15 (then the September peaks will trigger retraining)

- or monitor other features (seasonal assortment, new article influx),

which actually drift.
